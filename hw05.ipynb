### Problem 3 (35)

A smelter sits on the floor of a narrow valley that runs east–west. One
clear night there was almost no wind, and a **temperature inversion**
formed over the valley floor.

Air normally gets colder with height, so air warmed near the ground
rises and carries pollution up and away. On a clear, calm night the
ground cools quickly and chills the air just above it, which leaves a
pool of cold air sitting under warmer air. This is called an
**inversion**. The cold air is heavier than the air above it and cannot
rise, so the warmer air acts as a lid. That night the smelter’s SO$_2$
stayed in the pool of cold air on the valley floor. With almost no wind
to carry it out of the valley, its concentration built up through the
night.

The facility’s anemometer averaged 1 m/s from the west over the night,
but the wind was light and swung through every direction.

Twelve monitors on the valley floor recorded the night’s average SO$_2$
concentration. Their readings are in `data/receptors.csv`, with
positions measured from the stack: `downwind_km` is the distance east of
the stack (negative values are west of it), and `crosswind_m` is the
distance north.

The facility’s consultant modelled the night with the Gaussian plume
model below, using stability class F and the night’s average wind, 1 m/s
from the west. The stack’s effective height, 41 m including plume rise,
is known. Your job is to decide whether that model is fit to be used.

The stability classes run from A (strong daytime heating) through D
(neutral: overcast or windy) to F (a clear, calm night).

```julia
# Pasquill-Gifford coefficients for each stability class. Distances in km,
# sigmas in metres.
stability_coefs = Dict(
    "A" => (; a=213, c_near=440.8, d_near=1.941, f_near=9.27,
              c_far=459.7, d_far=2.094, f_far=-9.6),
    "B" => (; a=156, c_near=106.6, d_near=1.149, f_near=3.3,
              c_far=108.2, d_far=1.098, f_far=2),
    "C" => (; a=104, c_near=61, d_near=0.911, f_near=0,
              c_far=61, d_far=0.911, f_far=0),
    "D" => (; a=68, c_near=33.2, d_near=0.725, f_near=-1.7,
              c_far=44.5, d_far=0.516, f_far=-13),
    "E" => (; a=50.5, c_near=22.8, d_near=0.678, f_near=-1.3,
              c_far=55.4, d_far=0.305, f_far=-34),
    "F" => (; a=34, c_near=14.35, d_near=0.740, f_near=-0.35,
              c_far=62.6, d_far=0.180, f_far=-48.6))
class_names = ["A", "B", "C", "D", "E", "F"]

sigma_y(downwind_km, coef) = coef.a * downwind_km^0.894

function sigma_z(downwind_km, coef)
    if downwind_km < 1
        return coef.c_near * downwind_km^coef.d_near + coef.f_near
    else
        return coef.c_far * downwind_km^coef.d_far + coef.f_far
    end
end

# Ground-level concentration [µg/m³]. The wind blows along the downwind axis,
# so the model puts nothing upwind of the stack.
function ground_concentration(downwind_km, crosswind_m, coef;
                              emission=100, height=41, wind=1)
    if downwind_km <= 0.05
        return 0.0
    end
    spread_y = sigma_y(downwind_km, coef)
    spread_z = sigma_z(downwind_km, coef)
    if spread_z <= 0
        return 0.0
    end
    return (emission / (π * wind * spread_y * spread_z)) *
        exp(-crosswind_m^2 / (2 * spread_y^2)) *
        exp(-height^2 / (2 * spread_z^2)) * 1e6
end

monitors = CSV.read("data/receptors.csv", DataFrame)
```

```julia
# convert table above  
using Statistics

monitor_table = monitors 

monitor_downwind_km = monitor_table.downwind_km # east of stack in km
monitor_crosswind_m = monitor_table.crosswind_m  # north of stack in meter, not km 
observed = monitor_table.observed_ugm3 

number_of_monitors = length(observed) # Given that there are 12 
observed_mean = mean(observed) # Take mean of the final column, needed for P3.1  
```

#### Problem 3.1 (6)

Run the consultant’s model (class F, 1 m/s) at each monitor and compute
the RMSE against the observations. Report it both in µg/m³ and as a
percentage of the mean observed concentration.

On that number alone, would you say the model fits?

### Answer to 3.1
The RMSE against the observations is 2182.47 µg/m³ and 86.36 % of the mean observed concentration. 
Since the RMSE is very large, on the same order of magnitude of the observed concentration, the model **does not fit**. In addition, the modeled concentration shows a pattern of zero concentration for the first few monitors (since M01 to M05 are upwind) but then unstably increasing concentrations in the later monitors. This lead to the large RMSE, and the model is not a good fit.  



```julia
# For the specified class 
coeff_class_F = stability_coefs["F"]


modeled = zeros(number_of_monitors)
for i in 1:number_of_monitors
    modeled[i] = ground_concentration(monitor_downwind_km[i], monitor_crosswind_m[i], coeff_class_F; wind=1.0)
end

println(modeled)

# Get the error by taking the diff b/w model and observed
error = modeled .- observed # Need broadcasting 

# To find the RMSE 
square_error = (error).^2

mean1 = mean(square_error) # Can NOT name variable just using "mean"  
rmse = sqrt(mean1) 

# Step 4: RMSE as % of the mean observation
rmse_percent = 100 * rmse / observed_mean

println("RMSE is", round(rmse, digits = 2), " µg/m³")
println("RMSE in percentage is ", round(rmse_percent, digits = 2), " %")
```

Wind speed is the most uncertain input here, and the whole concentration

Keeping class F, fit the wind speed by minimising RMSE against the

### Answer to 3.2 

1. **The fitted wind speed is 1.3 m/s and the RMSE after calibration is about 2145.44 µg/m³**. 

2. Since the difference between the RMSE before and after the calibration is only about 37 µg/m³, the calibration **did not improve the fit meaningfully**, as the change of 37 is only about 1.7 % of the new RMSE. 

3. The wind speed can help scale the concentration by a factor of 1/u, which is a stretch or shrink in the x direction. However, wind speed cannot change the shape of the vertical and crosswind spread because these are controlled by the H, stability class, sigma z and y.  

As I learned in the OH with Professor, the issue with this problem's model is that the wind is very calm, at a speed of 1m/s, so that our Gussian plume model does NOT work because that violates our assumption. When we were deriving the Gaussian plume model, we assume downwind (in x direction) advection casued by the wind speed is much more significant than the turbulence in x direction, that is why we do not have a sigma x term. However, when the wind is calm, the turbulence in x direction would be more significant than the advenction, that's why our Gussian plume model fails to describe this situtaion. On the other hand, the advection-diffusion-turbulence model would always hold, if we were to derive that for this question with a very small wind speed close to 0, that new model would would then describe this situation corretly.  


> **One-variable search in Optim**
>
> `Optim.optimize(f, lower, upper)` minimises a function `f` of one
> number between two bounds. Here `f` can be
> `u -> model_rmse(stability_coefs["F"], u)`. `Optim.minimizer` returns
> the best value found and `Optim.minimum` the function’s value there.

```julia
function model_rmse(coef, wind_speed)
    # TODO: run the model at every monitor with these stability coefficients
    # and this wind speed, and return the RMSE against the observed
    # concentrations.
    modeled = zeros(number_of_monitors)

    for i in 1:number_of_monitors
        modeled[i] = ground_concentration(
            monitor_downwind_km[i], monitor_crosswind_m[i], coef; wind=wind_speed)
    end

    error_at_each_monitor = modeled .- observed 
    rmse_value = sqrt(mean(error_at_each_monitor.^2))
    return rmse_value

    #error("model_rmse has not been written yet -- replace this line")
end


# Test different wind speed to minimize RMSE 
RMSE_test = u -> model_rmse(stability_coefs["F"], u)

optimization = Optim.optimize(RMSE_test, 0.3, 20)
println(optimization)

wind_min_rmse = Optim.minimizer(optimization)   # best wind speed
min_rmse = Optim.minimum(optimization)     # RMSE at that wind speed

println(" ")
println(round(wind_min_rmse, digits=1), "m/s") # Wind speed that gives minimized RMSE 
println(min_rmse, "µg/m³") # RMSE after calibration 

println(rmse - min_rmse) # Difference in RMSE 
```

```julia
########## P3.3 ########

# 6 total classes 
fitted_wind_class = zeros(6)
fit_rmse_class = zeros(6)

# wind speed range to test 
lower_bound = 0.3
upper_bound = 20.0

for k in 1:6
    class_name = class_names[k] # from A to F, later for storing restuls in data  
    coeff_this_class= stability_coefs[class_name]

    optimization = Optim.optimize(u -> model_rmse(coeff_this_class, u), lower_bound, upper_bound)

    fitted_wind_class[k] = Optim.minimizer(optimization)

    fit_rmse_class[k] = Optim.minimum(optimization)
end

rmse_percent = 100 .* fit_rmse_class ./ observed_mean

table = DataFrame(stability_class = class_names, fitted_wind = fitted_wind_class, RMSE = fit_rmse_class, RMSE_percent = rmse_percent)

# output 
display(table)


# Use best class to compare w observed 
best_index = 4 # class D with the minimum RMSE in the "table" and D is the fourth one of the class 

best_class = class_names[best_index]
best_wind = fitted_wind_class[best_index]
best_coeff = stability_coefs[best_class]

println("Best class is ", best_class)
println("wind speed is ", round(best_wind, digits=2))
println("RMSE is ", round(fit_rmse_class[best_index], digits=2))

println(round(rmse_percent[best_index], digits=2))



### To get best class's modeled 
modelled_best = zeros(number_of_monitors)

for i in 1:number_of_monitors
    modelled_best[i] = ground_concentration(monitor_downwind_km[i], monitor_crosswind_m[i], best_coeff; wind= best_wind)
end

# Produce table with comparison to observed 
comparison_table = DataFrame(monitor = monitor_table.monitor, downwind_km = monitor_downwind_km, crosswind_m = monitor_crosswind_m,
    observed = observed, modelled = modelled_best, error = modelled_best .- observed)

comparison_table
```

You are asked to certify the consultant’s model as fit for deciding

### Answer to 3.4 


1. I would **not** certify it because the model has a very large RMSE and the monitors M01 to M05 (upwind ones) have modeled concentration of 0, which are very far off from the observed concentration of around 2000 µg/m³. 

2. Firstly, the assumption that the wind only carries SO2 in one direction (downwind) does not hold, because the actual setting has both SO2 concentration in different directions. Thus, **the plue model assumption of constant wind velocity and direction does not hold**. This is tied to the prompt saying the wind "swug through in every direction". Therefore, the dispersion would end up more like a blob instead of a plume, as Professor mentioned in OH. Secondly, the assumption that **wind >> dispersion in x direction does not hold** because the prompt mentioned that "wind is light" and  the observation data with nonzero upwind (monitor M01 - M05) concentrationsso, we know that dispersion would then matter as the wind is not strong enough to push the pollutant downwind faster than the dispersion happening along x direction (even impacting upwind locations/monitors). Thus, since our plume model was derived using the assumption that there is no sigma x (dispersion in x direction), it fails to describe a night like this one. 

3. **The plume model is just wrong for this scenario because the assumptions to apply our Gaussian plume model are not satisfied.** Yet, the comment on the calibrated model being "as good as it can be" is actually correct, since there is only so much to calibrate given the wind speed and stability case scenarios, as Professor mentioned in OH.
